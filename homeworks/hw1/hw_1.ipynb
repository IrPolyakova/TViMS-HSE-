```python
# 'Default' — технические победы (неявки), их выкидываем
res = results[results._map != 'Default'].drop_duplicates(['match_id', '_map']).copy()
res['date'] = pd.to_datetime(res.date)
res['year'] = res.date.dt.year
res['rounds'] = res.result_1 + res.result_2
res['ct_rounds'] = res.ct_1 + res.ct_2          # раунды, выигранные стороной CT (без овертаймов)
res['t_rounds'] = res.t_1 + res.t_2
res['reg_rounds'] = res.ct_rounds + res.t_rounds
res['ct_share'] = res.ct_rounds / res.reg_rounds

def team_view(df):
    """Каждая карта -> две строки, с точки зрения каждой из команд."""
    out = []
    for s in (1, 2):
        o = 3 - s
        out.append(pd.DataFrame({
            'date': df.date, 'year': df.year, 'match_id': df.match_id, 'map_name': df._map,
            'team': df[f'team_{s}'], 'opp': df[f'team_{o}'],
            'rounds_won': df[f'result_{s}'], 'rounds_lost': df[f'result_{o}'],
            # CT-раунды команды: выигранные = ct_s, проигранные = T-победы соперника в той же половине
            'ct_won': df[f'ct_{s}'], 'ct_lost': df[f't_{o}'],
            't_won': df[f't_{s}'], 't_lost': df[f'ct_{o}'],
            'win': (df.map_winner == s).astype(int),
            'rank_own': df[f'rank_{s}'], 'rank_opp': df[f'rank_{o}'],
        }))
    out = pd.concat(out, ignore_index=True)
    out['lrd'] = np.log(out.rank_opp) - np.log(out.rank_own)   # >0: мы выше в рейтинге HLTV
    return out

tm = team_view(res)
print(res.shape, tm.shape)
tm.head()
```

```python
# players: одна строка = игрок x матч, статистика по картам лежит "в ширину" (m1_*, m2_*, m3_*)
id_cols = ['date', 'player_name', 'player_id', 'team', 'opponent', 'country',
           'match_id', 'event_id', 'event_name', 'best_of']
stat_cols = ['kills', 'assists', 'deaths', 'hs', 'flash_assists', 'kast', 'kddiff', 'adr', 'fkdiff', 'rating',
             'kills_ct', 'deaths_ct', 'kddiff_ct', 'adr_ct', 'kast_ct', 'rating_ct',
             'kills_t', 'deaths_t', 'kddiff_t', 'adr_t', 'kast_t', 'rating_t']
parts = []
for k in (1, 2, 3):
    d = players[id_cols + [f'map_{k}'] + [f'm{k}_{c}' for c in stat_cols]].copy()
    d.columns = id_cols + ['map_name'] + stat_cols
    d['map_num'] = k
    parts.append(d)
pm = pd.concat(parts, ignore_index=True).dropna(subset=['map_name', 'kills'])
print('игрок x карта (все):', pm.shape)

res_cols = ['match_id', '_map', 'team_1', 'team_2', 'result_1', 'result_2', 'map_winner', 'rank_1', 'rank_2', 'rounds']
pm = pm.merge(res[res_cols].rename(columns={'_map': 'map_name'}), on=['match_id', 'map_name'], how='inner')
pm['side'] = np.select([pm.team == pm.team_1, pm.team == pm.team_2], [1, 2], 0)
pm = pm[pm.side > 0].copy()

pm['date'] = pd.to_datetime(pm.date)
pm['year'] = pm.date.dt.year
pm['rounds_won'] = np.where(pm.side == 1, pm.result_1, pm.result_2)
pm['round_share'] = pm.rounds_won / pm.rounds
pm['win'] = (pm.map_winner == pm.side).astype(int)
pm['rank_own'] = np.where(pm.side == 1, pm.rank_1, pm.rank_2)
pm['rank_opp'] = np.where(pm.side == 1, pm.rank_2, pm.rank_1)
pm['lrd'] = np.log(pm.rank_opp) - np.log(pm.rank_own)
pm['kpr'] = pm.kills / pm.rounds
pm['dpr'] = pm.deaths / pm.rounds
pm['apr'] = pm.assists / pm.rounds
pm['fkpr'] = pm.fkdiff / pm.rounds              # разница первых убийств на раунд
pm['hs_share'] = pm.hs / pm.kills.replace(0, np.nan)

# оставляем только карты, где у команды ровно 5 игроков в данных
pm['n_in_team'] = pm.groupby(['match_id', 'map_name', 'team']).player_id.transform('size')
pm = pm[pm.n_in_team == 5].drop(columns=['team_1', 'team_2', 'result_1', 'result_2', 'rank_1', 'rank_2']).reset_index(drop=True)

# понадобится только матчевый rating игроков — остальное освобождаем
hist = players[['date', 'player_id', 'match_id', 'rating']].copy()
del players, parts, d
_ = gc.collect()
print('после склейки с results:', pm.shape, '| игроков:', pm.player_id.nunique(), '| матчей:', pm.match_id.nunique())
pm[['rating', 'kpr', 'dpr', 'apr', 'kast', 'adr', 'fkpr', 'hs_share', 'round_share']].describe().T.round(3)
```
